```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Настраиваем размер графиков
plt.rcParams['figure.figsize'] = (10, 5)
plt.style.use('default')

# 1. Загружаем чистую таблицу
df = pd.read_csv('../data/processed/matches_clean.csv')

# Создаем папку для графиков, если ее нет
os.makedirs('../reports/figures', exist_ok=True)

print("Таблица загружена. Начинаем строить графики...")

# --- ГРАФИК 1: Исходы матчей (H, D, A) ---
plt.figure()
df['result'].value_counts().reindex(['H','D','A']).plot(kind='bar', color=['#2e7d32','#9e9e9e','#c62828'])
plt.title('Исходы матчей Ла Лиги (2013-2026)')
plt.xlabel('H - победа хозяев, D - ничья, A - победа гостей')
plt.ylabel('Число матчей')
plt.xticks(rotation=0)
plt.savefig('../reports/figures/01_results.png', dpi=120, bbox_inches='tight')
plt.close()
print("График 1 готов")

# --- ГРАФИК 2: Среднее число голов по сезонам ---
plt.figure()
goals = df.groupby('season').apply(lambda x: (x['home_goals'] + x['away_goals']).mean())
goals.plot(kind='line', marker='o', color='#1565c0')
plt.title('Среднее число голов за матч по сезонам')
plt.xlabel('Сезон')
plt.ylabel('Голов за матч')
plt.grid(True, alpha=0.3)
plt.savefig('../reports/figures/02_goals_by_season.png', dpi=120, bbox_inches='tight')
plt.close()
print("График 2 готов")

# --- ГРАФИК 3: Доля побед хозяев по сезонам ---
plt.figure()
home_win = df.groupby('season').apply(lambda x: (x['result']=='H').mean())
home_win.plot(kind='line', marker='s', color='#2e7d32')
plt.title('Доля побед хозяев по сезонам')
plt.xlabel('Сезон')
plt.ylabel('Доля побед хозяев')
plt.ylim(0.3, 0.6)
plt.grid(True, alpha=0.3)
plt.savefig('../reports/figures/03_home_wins.png', dpi=120, bbox_inches='tight')
plt.close()
print("График 3 готов")

# --- ГРАФИК 4: Распределение голов за матч ---
plt.figure()
total_goals = df['home_goals'] + df['away_goals']
total_goals.hist(bins=range(0, 12), color='#6a1b9a', edgecolor='white')
plt.title('Распределение числа голов за матч')
plt.xlabel('Голов за матч')
plt.ylabel('Число матчей')
plt.savefig('../reports/figures/04_goals_dist.png', dpi=120, bbox_inches='tight')
plt.close()
print("График 4 готов")

# --- ГРАФИК 5: Результаты Барселоны по сезонам ---
plt.figure()
barca = df[(df['home']=='Barcelona') | (df['away']=='Barcelona')].copy()
def barca_result(row):
    if row['home']=='Barcelona':
        return 'H' if row['result']=='H' else ('A' if row['result']=='A' else 'D')
    else:
        return 'A' if row['result']=='H' else ('H' if row['result']=='A' else 'D')
barca['barca_res'] = barca.apply(barca_result, axis=1)
barca.groupby(['season','barca_res']).size().unstack(fill_value=0).reindex(columns=['H','D','A'], fill_value=0).plot(kind='bar', stacked=True, color=['#2e7d32','#9e9e9e','#c62828'])
plt.title('Результаты «Барселоны» по сезонам')
plt.xlabel('Сезон')
plt.ylabel('Число матчей')
plt.legend(title='Исход для Барсы')
plt.savefig('../reports/figures/05_barca.png', dpi=120, bbox_inches='tight')
plt.close()
print("График 5 готов")

# --- ГРАФИК 6: Топ-10 команд по числу побед ---
plt.figure()
wins = pd.concat([df[df['result']=='H']['home'], df[df['result']=='A']['away']]).value_counts().head(10)
wins.sort_values().plot(kind='barh', color='#ef6c00')
plt.title('Топ-10 команд по числу побед (2013-2026)')
plt.xlabel('Побед')
plt.savefig('../reports/figures/06_top_wins.png', dpi=120, bbox_inches='tight')
plt.close()
print("График 6 готов")

print("ВСЁ ГОТОВО! Проверь папку reports/figures")
```
